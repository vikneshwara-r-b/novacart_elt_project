### Control Table Setup

```python
# Remove existing widgets (safe to re-run)
dbutils.widgets.removeAll()

# Create widgets for catalog and schema names
dbutils.widgets.text("catalog_name", "novacart_catalog", "1. Catalog Name")
dbutils.widgets.text("bronze_schema", "bronze_schema", "2. Bronze Schema Name")
dbutils.widgets.text("silver_schema", "silver_schema", "3. Silver Schema Name")
dbutils.widgets.text("gold_schema", "gold_schema", "4. Gold Schema Name")
```

#### Bronze control table
This table stores the **watermark** for each source table.
It helps the pipeline remember:
- the latest timestamp already processed
- the latest primary key processed at that timestamp
- how many rows were written in the latest run
This is what makes the Bronze load incremental and rerun-safe.

```python
%sql
CREATE TABLE IF NOT EXISTS IDENTIFIER(:catalog_name || '.' || :bronze_schema || '.ingestion_control')(
    layer string,
    table_name string,
    ts_col string,
    pk_col string,
    last_successful_ts timestamp,
    last_successful_pk bigint,
    last_run_id string,
    rows_written bigint,
    run_status string,
    updated_at timestamp
)
USING DELTA
```

#### Silver control table
This table stores the latest Silver processing state for each entity

It helps us track:
- the latest Bronze run already processed by Silver
- the latest Bronze ingestion timestamp already processed
- How many rows were merged in the latest Silver un

```python
%sql
CREATE TABLE IF NOT EXISTS IDENTIFIER(:catalog_name || '.' || :silver_schema || '.processing_control')(
    layer string,
    entity_name string,
    last_processed_bronze_run_id string,
    last_processed_bronze_ingested_at timestamp,
    rows_merged bigint,
    run_status string,
    silver_run_id string,
    updated_at timestamp
)
USING DELTA
```

#### Gold control table
This table stores the latest Gold processing state.

It tells Gold:
- which Silver data was processed last time
- how many Gold rows were merged in the last run

```python
%sql
CREATE TABLE IF NOT EXISTS IDENTIFIER(:catalog_name || '.' || :gold_schema || '.processing_control')(
    layer string,
    entity_name string,
    last_processed_silver_run_id string,
    last_processed_silver_run_ts timestamp,
    rows_merged bigint,
    run_status string,
    gold_run_id string,
    updated_at timestamp
)
USING DELTA
```
