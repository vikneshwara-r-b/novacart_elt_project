## Step 1: Configure Input Widgets

```python
# Remove existing widgets (safe to re-run)
dbutils.widgets.removeAll()
 
# Connection details
dbutils.widgets.text("catalog_name", "", "1. Catalog Name which have to be created")
dbutils.widgets.text("bronze_schema_name", "", "2. Bronze Schema Name")
dbutils.widgets.text("silver_schema_name", "", "3. Silver Schema Name")
dbutils.widgets.text("gold_schema_name", "", "4. Gold Schema Name")
```

## Step 2: Read & Validate Widget Values

```python
# Populate python variables for catalog creation information
catalog_name   = dbutils.widgets.get("catalog_name").strip()
bronze_schema_name = dbutils.widgets.get("bronze_schema_name").strip()
silver_schema_name = dbutils.widgets.get("silver_schema_name").strip()
gold_schema_name = dbutils.widgets.get("gold_schema_name").strip()
```

```python
# Validate all required fields are provided
required = {
    "catalog_name":    catalog_name,
    "bronze_schema_name": bronze_schema_name,
    "silver_schema_name": silver_schema_name,
    "gold_schema_name": gold_schema_name,
}
 
missing = [k for k, v in required.items() if not v]
 
if missing:
    raise ValueError(f"The following required widget values are missing or empty: {missing}")
 
print("All widget values provided.")
print(f"Catalog Name: {catalog_name}")
print(f"Bronze Schema Name: {bronze_schema_name}")
print(f"Silver Schema Name: {silver_schema_name}")
print(f"Gold Schema Name: {gold_schema_name}")
```

## Step 3: Create Unity Catalog and Dependent Schema Names
Uses `CREATE CATALOG IF NOT EXISTS` so the notebook is safe to re-run.

```python
# Catalog creation
create_catalog_sql = f"""
CREATE CATALOG IF NOT EXISTS `{catalog_name}`
COMMENT 'Catalog created for Novacart Case Study'
"""
 
spark.sql(create_catalog_sql)
print(f"Catalog '{catalog_name}' created (or already exists).")
```

```python
# Bronze schema creation
create_bronze_schema_sql = f"""
CREATE SCHEMA IF NOT EXISTS `{catalog_name}`.`{bronze_schema_name}`
COMMENT 'Bronze Schema in custom {catalog_name}'
"""
spark.sql(create_bronze_schema_sql)
print(f"Bronze schema '{bronze_schema_name}' created (or already exists).")
```

```python
# Silver schema creation
create_silver_schema_sql = f"""
CREATE SCHEMA IF NOT EXISTS `{catalog_name}`.`{silver_schema_name}`
COMMENT 'Silver Schema in custom {catalog_name}'
"""
spark.sql(create_silver_schema_sql)
print(f"Silver schema '{silver_schema_name}' created (or already exists).")
```

```python
# Gold schema creation
create_gold_schema_sql = f"""
CREATE SCHEMA IF NOT EXISTS `{catalog_name}`.`{gold_schema_name}`
COMMENT 'Gold Schema in custom {catalog_name}'
"""
spark.sql(create_gold_schema_sql)
print(f"Gold schema '{gold_schema_name}' created (or already exists).")
```
